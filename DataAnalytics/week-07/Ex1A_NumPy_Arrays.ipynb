# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA: 
    for y in x: 
        print(y * 2)
```

The first loop works on all of the rows and print shows 3 doubled values. The second loop handles single numbers so each print shows one value.

```python
for x in arrA: 
    for y in x: 
        loop_result = x * 2
        print(loop_result)
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
# ValueError all input arrays must have the same shape
# Joined_array_1 is 4x3 and joined_array_2 is 2x6
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
# ValueError arrays must match in every dimension except the one being joined.
# Joining side by side needs the same number of rows.
```

```python
h = np.hstack((arrA, arrA))
# hstack joins side by side horizontally
v = np.vstack((arrA, arrA))
# vstack joins top to bottom vertically
d = np.dstack((arrA, arrA))
# dstack goes off depth and the layers along the third axis
# Each position holds a pair of matching values

print(h)
print(v)
print(d)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# 2 arrays each 2 dimensional
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# 3 arrays each 2 dimensional: 1 with 2 rows then 2 with 1 row
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7 
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
