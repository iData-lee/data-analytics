# Exercise 2.C Creating and Calling functions

## Lab 1. Working with defining custom functions

```python
def fav_things(name, movie, music):
    return f'Name: {name}\nFavorite Movie: {movie}\nFavorite Musician/Band: {music}'
```

```python
print(fav_things('Lale', 'Life of Pi', 'Skrillex'))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here('I joined the Data Analytics pathway because I wanted a versatile career that is not only useful in all fields but a career I can love and take care of my family.'))
```

```python
def fav_place(place, why):
    return f'One of my favorite places is {place} becuase {why}.'
```

```python
print(fav_place('the beach', 'its tranquil and calming especially at night' ))
```

```python
print(fav_place('the mountains', 'I love the fresh air and the view of the sunset from the top of a mountain.'))
```
