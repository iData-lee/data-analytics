# Exercise 2.C Lab 3. Tax Functions
This notebook defines functions that calculate Social Security, Medicare, and federal withholding taxes froma gross pay amount.

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, code):
    if code == 0:
        rate = 0.23
    elif code == 1:
        rate = 0.21
    elif code == 2:
        rate = 0.195
    elif code == 3:
        rate = 0.185 
    else:
        # code 4 or more: 18% less 0.5% for each code over 4
        rate = 0.18 - (code - 4) * 0.005
    return gross_pay * rate
```

```python
print(f'Social Security: ${get_soc_sec_tax(750):.2f}')
print(f'Medicare: ${get_medicare_tax(750):.2f}')
print(f'Federal: ${get_federal_tax(750, 0):.2f}')
```

```python
print(f'Social Security: ${get_soc_sec_tax(1550):.2f}')
print(f'Medicare: ${get_medicare_tax(1550):.2f}')
print(f'Federal: ${get_federal_tax(1550, 2):.2f}')
```
