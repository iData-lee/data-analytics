## Lab 2. Parsing strings - part codes

```python
supplier:productcode-size
```

```python
# Returns all characters before the colon
def get_supplier_code(code):
    colon = code.find(':')
    return code[:colon]
```

```python
print(get_supplier_code('ACME:123-L'))
```

```python
# Returns all characters between the colon and the dash
def get_product_num(code):
    colon = code.find(':')
    dash = code.find('-')
    return code[colon + 1:dash]
```

```python
print(get_product_num('ACME:123-L'))
```

```python
# Returns all characters after the dash
def get_size(code):
    dash = code.find('-')
    return code[dash + 1:]
```

```python
print(get_size('ACME:123-L'))
```

```python
part1 = 'ACME:123-L'
part2 = 'DI:12345-M'
part3 = 'ACE:1-12'
```

```python
print(f'For {part1}, the supplier is {get_supplier_code(part1)}. Part number {get_product_num(part1)} is available in size {get_size(part1)}.')
```

```python
print(f'For {part2}, the supplier is {get_supplier_code(part2)}. Part number {get_product_num(part2)} is available in size {get_size(part2)}.')
```

```python
print(f'For {part3}, the supplier is {get_supplier_code(part3)}. Part number {get_product_num(part3)} is available in size {get_size(part3)}.')
```
